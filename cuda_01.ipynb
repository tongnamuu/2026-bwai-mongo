```python
%%writefile kernel.cu
#include <stdio.h>

__global__ void helloCUDA() {
    printf("Hello World from GPU thread %d\n", threadIdx.x);
}

int main() {
    helloCUDA<<<1, 10>>>();
    cudaDeviceSynchronize();
    return 0;
}
```

```python
!nvcc kernel.cu -o app && ./app
```
